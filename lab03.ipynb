```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 8))

# Set limits for a square plot to ensure shapes are not distorted
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)

# Draw the first circle
circle1 = plt.Circle((2, 5), 1.0, color='blue', alpha=0.7)
ax.add_patch(circle1)

# Draw the second circle
circle2 = plt.Circle((6, 5), 1.0, color='red', alpha=0.7)
ax.add_patch(circle2)

# Draw the rectangle in the middle
# The coordinates are (x, y) of the bottom-left corner, width, height
rectangle = plt.Rectangle((3, 4), 2, 8, color='green', alpha=0.7)
ax.add_patch(rectangle)

# Add title and remove axes ticks for a cleaner look
ax.set_title('Two Circles and a Rectangle')
ax.set_aspect('equal', adjustable='box') # Ensure circles appear as circles
ax.axis('off') # Hide the axes

plt.show()

from google.colab import drive
drive.mount('/content/drive')
```
